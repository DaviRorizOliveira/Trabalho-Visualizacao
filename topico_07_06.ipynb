```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, Rectangle
import warnings
warnings.filterwarnings("ignore")

# Cores utilizadas
CORES = {
    "laranja": "#E07A3D",
    "azul": "#2E5A88",
    "verde": "#3A7D5C",
    "roxo": "#6B4C9A",
    "cinza": "#5A5A5A",
    "cinza_claro": "#E8E8E8",
    "fundo": "#FAFAFA",
    "texto": "#222222",
}
```

# Matrizes e Listas

O exemplo escoilhido foi um gráfico em formato de matriz, cujo objetivo é comparar sistematicamente as características técnicas de diferentes categorias de sistemas EEG comerciais com a proposta ECOEEG, permitindo identificar semelhanças, diferenças, vantagens e limitações entre as soluções existentes.

```python
def matriz_comparativa():
    fig, ax = plt.subplots(figsize=(16, 9))
    fig.patch.set_facecolor(CORES["fundo"])
    ax.set_xlim(0, 16)
    ax.set_ylim(0, 10)
    ax.axis("off")

    ax.text(8, 9.5, "Matriz comparativa: sistemas EEG comerciais por faixa de preço vs. ECOEEG", ha="center", fontsize=15, fontweight="bold", color=CORES["texto"])

    # Geometria: coluna de rótulos à esquerda + 4 colunas de dados
    x_rotulo = 0.5
    larg_col = 2.8
    colunas = [
        (4.9,  "Alto custo\n(> US$ 10.000)", CORES["azul"]),
        (7.8,  "Médio custo\n(US$ 1.000 – 10.000)", CORES["roxo"]),
        (10.7, "Baixo custo comerciais\n(< US$ 1.000)", CORES["verde"]),
        (13.6, "ECOEEG\n(proposta do artigo)", CORES["laranja"]),
    ]

    # Cabeçalhos das colunas
    for x, titulo, cor in colunas:
        ax.add_patch(FancyBboxPatch((x - larg_col/2, 7.85), larg_col, 0.8, boxstyle="round,pad=0.05,rounding_size=0.15", facecolor=cor, edgecolor="none"))
        ax.text(x, 8.25, titulo, ha="center", va="center", fontsize=9.5, fontweight="bold", color="white", linespacing=1.2)

    ax.text(x_rotulo, 8.25, "Critério", ha="left", va="center", fontsize=10, fontweight="bold", color=CORES["texto"])

    criterios = [
        ("Nº de canais",
         "8 – 256 canais",
         "8 – 64 canais",
         "1 – 14 canais",
         "4 canais ativos\n+ bias + ref. na orelha"),
        ("Tipo de eletrodo",
         "Ativos ou passivos;\númidos ou secos",
         "Ativos ou passivos;\númidos, secos ou semi-secos",
         "Passivos; secos,\nsemi-secos ou úmidos",
         "Ativos úmidos (Ag/AgCl\n+ pré-amplificador)"),
        ("Resolução do ADC",
         "16 – 24 bits",
         "16 – 24 bits",
         "12 – 24 bits",
         "24 bits (ADS1299-4)"),
        ("Comunicação sem fio",
         "Parcial\n(3 de 5 sem conexão sem fio)",
         "Quase todos\n(8 de 9)",
         "Sim (todos)",
         "Sim (BLE)"),
        ("Certificações",
         "CE / FDA em parte\n(ex.: CGX, actiChamp)",
         "CE / FDA em 4 de 9",
         "Nenhuma informada",
         "Uso não clínico:\nsem CE/FDA"),
        ("Autonomia da bateria",
         "2 – 16 h",
         "5 – 24 h",
         "4 – 9 h",
         "> 35 h (medido)"),
        ("Montagem dos eletrodos",
         "Muitos eletrodos a\npreparar (ex.: actiChamp)",
         "Muitos eletrodos a\npreparar (ex.: g.Nautilus PRO)",
         "Faixa rígida,\nposições fixas",
         "Faixa elástica,\nposições reconfiguráveis (10-10)"),
    ]

    y0 = 7.3
    passo = 0.88
    for i, (crit, *valores) in enumerate(criterios):
        y = y0 - i * passo
        if i % 2 == 0:
            ax.add_patch(Rectangle((0.3, y - 0.4), 15.1, 0.8, facecolor="#F0F0F0", edgecolor="none", alpha=0.6))

        ax.text(x_rotulo, y, crit, ha="left", va="center", fontsize=9.5, fontweight="bold", color=CORES["texto"])
        for (x, _, _), valor in zip(colunas, valores):
            ax.text(x, y, valor, ha="center", va="center", fontsize=8.5, color=CORES["cinza"], linespacing=1.2)

    plt.tight_layout()
    plt.savefig("figs/topico_07_06_01.png", dpi=300, bbox_inches="tight", facecolor=fig.get_facecolor())
    plt.show()

matriz_comparativa()
```
