```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, Rectangle
import textwrap
import warnings
warnings.filterwarnings("ignore")

# Cores utilizadas
CORES = {
    "laranja": "#E07A3D",
    "azul": "#2E5A88",
    "verde": "#3A7D5C",
    "roxo": "#6B4C9A",
    "cinza": "#5A5A5A",
    "cinza_claro": "#E8E8E8",
    "fundo": "#FAFAFA",
    "texto": "#222222",
}
```

# Citações

O objetivo desse gráfico é apresentar os principais fundamentos teóricos e informações da literatura científica que contextualizam a utilização da eletroencefalografia (EEG), destacando sua relevância, aplicações e evolução tecnológica a partir de contribuições de diferentes pesquisadores. Cada citação foi posta em um "cartão", que inclui a citação e o autor como informações, e uma aspas duplas, barra lateral e caixa envolto das informações como forma de decoração, além disso, foi escolhido uma cor diferente para cada citação.

```python
# Citações

def cartoes_citacoes():
    citacoes = [
        {   # Definição e relevância do EEG
            "texto": "A eletroencefalografia (EEG) é uma técnica não invasiva amplamente utilizada para adquirir sinais cerebrais, desempenhando um papel crucial na neurociência, no diagnóstico clínico e na pesquisa cognitiva.",
            "fonte": "S. R. Liyanage e C. Bhatt (2020)",
            "cor": CORES["laranja"]
        },
        {   # Aplicações do EEG e interferências no sinal
            "texto": "O EEG é comumente empregado no diagnóstico de distúrbios neurológicos, estudos do sono, avaliação de carga cognitiva e pesquisas neurofisiológicas, como interfaces cérebro-computador (BCI).",
            "fonte": "J. R. Wolpaw e E. W. Wolpaw (2012)",
            "cor": CORES["azul"]
        },
        {   # Integração com tecnologias assistivas
            "texto": "À medida que esse ramo da ciência continua a avançar, maior ênfase é dada a aplicações práticas, por exemplo, vinculando tais sistemas a tecnologias assistivas existentes.",
            "fonte": "J. D. R. Millán et al. (2010)",
            "cor": CORES["verde"]
        },
        {   # Faixas de preço de sistemas comerciais de EEG
            "texto": "Os dispositivos são categorizados em três faixas de preço: alta (acima de US$ 10.000), média (US$ 1.000–10.000) e baixa (abaixo de US$ 1.000).",
            "fonte": "C. He et al. (2023)",
            "cor": CORES["roxo"]
        },
    ]

    fig, axes = plt.subplots(2, 2, figsize = (13, 9))
    fig.patch.set_facecolor(CORES["fundo"])
    axes = axes.flatten()

    for ax, cit in zip(axes, citacoes):
        ax.set_xlim(0, 10)
        ax.set_ylim(0, 6)
        ax.axis("off")

        # Fundo do cartão
        box = FancyBboxPatch((0.3, 0.4), 9.4, 5.2, boxstyle = "round,pad = 0.15,rounding_size = 0.3", facecolor = "white", edgecolor = cit["cor"], linewidth = 2.5)
        ax.add_patch(box)

        # Barra lateral colorida
        ax.add_patch(Rectangle((0.3, 0.4), 0.25, 5.2, facecolor = cit["cor"], edgecolor = "none"))

        # Aspas decorativas
        ax.text(0.9, 5.1, "“", fontsize = 42, color = cit["cor"], va = "top", fontweight = "bold", alpha = 0.7)

        # Texto da citação
        wrapped = textwrap.fill(cit["texto"], width = 48)
        ax.text(1.1, 3.6, wrapped, ha = "left", va = "center", fontsize = 10.5, color = CORES["texto"], linespacing = 1.45)

        # Tema e fonte
        ax.text(1.1, 0.85, cit["fonte"], ha = "left", va = "center", fontsize = 8.5, color = CORES["cinza"], style = "italic")

    plt.savefig("figs/topico_07_03_01.png", dpi = 300, bbox_inches = "tight", facecolor = plt.gcf().get_facecolor())
    plt.show()

cartoes_citacoes()
```
